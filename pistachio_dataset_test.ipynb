# Pistachio Dataset Test

This notebook checks the dataset for scale, format and representation


```python
from pathlib import Path
import numpy as np
from PIL import Image

RESOURCE_DIR_KIRMIZI = Path("dataset/Pistachio_Image_Dataset/Pistachio_Image_Dataset/Kirmizi_Pistachio")
RESOURCE_DIR_SIIRT = Path("dataset/Pistachio_Image_Dataset/Pistachio_Image_Dataset/Siirt_Pistachio")
# Gather all image files (common extensions)
image_extensions = {'.jpg', '.jpeg', '.png', '.bmp', '.tiff'}
image_paths = (
    [p for p in RESOURCE_DIR_KIRMIZI.iterdir() if p.suffix.lower() in image_extensions] +
    [p for p in RESOURCE_DIR_SIIRT.iterdir() if p.suffix.lower() in image_extensions]
)

all_jpg = True
all_600x600 = True
issues = []

for img_path in image_paths:
    # Check extension
    if img_path.suffix.lower() not in ('.jpg', '.jpeg'):
        all_jpg = False
        issues.append(f"{img_path.name} → not a JPG (is {img_path.suffix})")
        continue  # skip PIL open to save time if you only care about JPG check

    # Check dimensions
    with Image.open(img_path) as img:
        width, height = img.size
        if (width, height) != (600, 600):
            all_600x600 = False
            issues.append(f"{img_path.name} → {width}×{height} (not 600×600)")

# Results
if all_jpg and all_600x600:
    print(f"All {len(image_paths)} images are JPGs and 600×600.")
else:
    print(f"Found {len(issues)} issue(s):")
    for msg in issues:
        print(f"  • {msg}")
    print(f"\nSummary: All JPGs? {all_jpg}  |  All 600×600? {all_600x600}")
```
